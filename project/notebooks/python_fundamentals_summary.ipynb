### 1. Create a Small ETF Example

```python
import numpy as np
import pandas as pd

from src.utils import require_columns, summarize_missing

prices = pd.DataFrame({
    'ticker': ['SPY', 'SPY', 'TLT', 'TLT'],
    'date': pd.to_datetime(['2026-01-02', '2026-01-05', '2026-01-02', '2026-01-05']),
    'close': [680.0, 684.0, 89.0, 88.5],
})
require_columns(prices, ['ticker', 'date', 'close'])
prices
```

### 2. Calculate and Summarize Returns

```python
prices = prices.sort_values(['ticker', 'date'])
prices['return_1d'] = prices.groupby('ticker')['close'].pct_change(fill_method=None)
average_return = prices.groupby('ticker')['return_1d'].mean()
annualized_example = np.sqrt(252) * prices['return_1d'].std()

print('Average return by ticker:')
display(average_return)
print(f'Annualized volatility example: {annualized_example:.4f}')
display(summarize_missing(prices))
```
