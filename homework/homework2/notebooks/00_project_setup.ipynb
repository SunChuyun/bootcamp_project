# Environment & Config Check

This notebook verifies the Python environment, local configuration, and a basic NumPy operation.

### 1. Verify the Environment

The interpreter and required packages should come from `bootcamp_env`.

```python
import sys
from pathlib import Path

import numpy as np
from dotenv import load_dotenv

from src.config import get_key

PROJECT_ROOT = Path.cwd()
print("Python version:", sys.version.split()[0])
print("Interpreter:", sys.executable)
print("Project root:", PROJECT_ROOT)
```

### 2. Load Local Configuration

The `.env` file stores local values and is excluded from Git.

```python
ENV_FILE = PROJECT_ROOT / ".env"
load_dotenv(dotenv_path=ENV_FILE)

api_key_present = get_key("API_KEY") is not None
data_dir = get_key("DATA_DIR", "./data")

print(".env loaded:", ENV_FILE.exists())
print("API_KEY present:", api_key_present)
print("DATA_DIR:", data_dir)
```

### 3. NumPy Check

A vectorized operation confirms that NumPy is installed and working.

```python
values = np.array([1, 2, 3, 4, 5])
scaled_values = values * 10

print("Array:", values)
print("Mean:", values.mean())
print("Vectorized * 10:", scaled_values)
```

- `requirements.txt` records the installed package versions.
- The NumPy output verifies a simple vectorized calculation.
